```python
output_dir = Path("python_analysis_outputs")
chart_dir = output_dir / "charts"
output_dir.mkdir(exist_ok=True)
chart_dir.mkdir(exist_ok=True)

sns.set_theme(style="whitegrid")

def save_chart(filename):
    plt.tight_layout()
    plt.savefig(chart_dir / filename, dpi=200, bbox_inches="tight")
    plt.show()
    plt.close()


# 1. Profit by category
category_summary = df.groupby("Category", as_index=False).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Units_Sold=("Quantity", "sum")
)

category_summary["Profit_Margin_Percent"] = (
    category_summary["Total_Profit"]
    / category_summary["Total_Sales"].replace(0, np.nan)
    * 100
)

plt.figure(figsize=(8, 4.5))
ax = sns.barplot(
    data=category_summary.sort_values("Total_Profit"),
    x="Total_Profit", y="Category", color="steelblue"
)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_title("Profit by Category")
ax.set_xlabel("Total Profit ($)")
ax.set_ylabel("")
save_chart("01_profit_by_category.png")


# 2. Profit by region
region_summary = df.groupby("Region", as_index=False).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
)

region_summary["Profit_Margin_Percent"] = (
    region_summary["Total_Profit"]
    / region_summary["Total_Sales"].replace(0, np.nan)
    * 100
)

plt.figure(figsize=(8, 4.5))
ax = sns.barplot(
    data=region_summary.sort_values("Total_Profit"),
    x="Total_Profit", y="Region", color="cornflowerblue"
)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_title("Profit by Region")
ax.set_xlabel("Total Profit ($)")
ax.set_ylabel("")
save_chart("02_profit_by_region.png")


# 3. Discount impact on profit
conditions = [
    df["Discount"].eq(0),
    df["Discount"].gt(0) & df["Discount"].le(0.10),
    df["Discount"].gt(0.10) & df["Discount"].le(0.20),
    df["Discount"].gt(0.20) & df["Discount"].le(0.30),
    df["Discount"].gt(0.30) & df["Discount"].le(0.40)
]
labels = [
    "0% - No discount", "1-10%", "11-20%",
    "21-30%", "31-40%"
]

df["Discount_Band"] = np.select(
    conditions, labels, default="Above 40%"
)
df.loc[df["Discount"].isna(), "Discount_Band"] = "Missing"

discount_summary = df.groupby("Discount_Band", as_index=False).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Transaction_Lines=("Order ID", "size")
)

band_order = labels + ["Above 40%", "Missing"]
discount_summary["Sort_Order"] = discount_summary["Discount_Band"].map(
    {band: i for i, band in enumerate(band_order)}
)
discount_summary = discount_summary.sort_values("Sort_Order")

plt.figure(figsize=(10, 5))
ax = sns.barplot(
    data=discount_summary,
    x="Discount_Band", y="Total_Profit", color="steelblue"
)
ax.axhline(0, color="black", linewidth=1)
ax.set_title("Profit by Discount Band")
ax.set_xlabel("Discount Band")
ax.set_ylabel("Total Profit ($)")
plt.xticks(rotation=25, ha="right")
save_chart("03_profit_by_discount_band.png")


# 4. Seasonal demand by category
season_data = df.loc[df["Order Date"].notna()].copy()
month = season_data["Order Date"].dt.month

season_data["Season"] = np.select(
    [
        month.isin([12, 1, 2]),
        month.isin([3, 4, 5]),
        month.isin([6, 7, 8])
    ],
    ["Winter", "Spring", "Summer"],
    default="Autumn"
)

season_order = ["Winter", "Spring", "Summer", "Autumn"]
season_data["Order_Year"] = season_data["Order Date"].dt.year

seasonal_category_summary = season_data.groupby(
    ["Season", "Category"], as_index=False
).agg(
    Units_Sold=("Quantity", "sum"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

season_pivot = seasonal_category_summary.pivot(
    index="Season", columns="Category", values="Units_Sold"
).reindex(season_order).fillna(0)

ax = season_pivot.plot(
    kind="bar", stacked=True, figsize=(9, 5)
)
ax.set_title("Seasonal Demand by Category")
ax.set_xlabel("Season")
ax.set_ylabel("Units Sold")
ax.legend(title="Category")
save_chart("04_seasonal_demand_by_category.png")


# 5. Compare seasonal demand across years
seasonal_year_summary = season_data.groupby(
    ["Order_Year", "Season"], as_index=False
).agg(
    Units_Sold=("Quantity", "sum"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

year_pivot = seasonal_year_summary.pivot(
    index="Order_Year", columns="Season", values="Units_Sold"
).reindex(columns=season_order)

ax = year_pivot.plot(marker="o", figsize=(9, 5))
ax.set_title("Seasonal Demand Across Years")
ax.set_xlabel("Order Year")
ax.set_ylabel("Units Sold")
ax.legend(title="Season")
save_chart("05_seasonal_demand_by_year.png")


# 6. Profit by shipping mode
shipping_summary = df.groupby("Ship Mode", as_index=False).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
)

shipping_summary["Profit_Margin_Percent"] = (
    shipping_summary["Total_Profit"]
    / shipping_summary["Total_Sales"].replace(0, np.nan)
    * 100
)

plt.figure(figsize=(8, 4.5))
ax = sns.barplot(
    data=shipping_summary.sort_values("Total_Profit"),
    x="Total_Profit", y="Ship Mode", color="cornflowerblue"
)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_title("Profit by Shipping Mode")
ax.set_xlabel("Total Profit ($)")
ax.set_ylabel("")
save_chart("06_profit_by_shipping_mode.png")


# Save summary tables for your report and further analysis
overall_summary = pd.DataFrame([{
    "Transaction_Rows": len(df),
    "Total_Orders": df["Order ID"].nunique(),
    "Total_Sales": df["Sales"].sum(),
    "Total_Profit": df["Profit"].sum(),
    "Profit_Margin_Percent": (
        df["Profit"].sum() / df["Sales"].sum() * 100
        if df["Sales"].sum() else np.nan
    )
}])

overall_summary.to_csv(output_dir / "overall_kpis.csv", index=False)
category_summary.to_csv(output_dir / "category_performance.csv", index=False)
region_summary.to_csv(output_dir / "region_performance.csv", index=False)
discount_summary.to_csv(output_dir / "discount_performance.csv", index=False)
seasonal_category_summary.to_csv(
    output_dir / "seasonal_category_performance.csv", index=False
)
seasonal_year_summary.to_csv(
    output_dir / "seasonal_by_year.csv", index=False
)
shipping_summary.to_csv(
    output_dir / "shipping_mode_performance.csv", index=False
)

print("Python chart generation completed.")
print("Charts and summary tables saved in:", output_dir.resolve())
display(overall_summary)
```
